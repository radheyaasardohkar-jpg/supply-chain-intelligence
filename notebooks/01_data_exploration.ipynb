```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px

# Load dataset
df = pd.read_csv("../data/raw/retail_store_inventory.csv")

# Convert Date to datetime
df["Date"] = pd.to_datetime(df["Date"])

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
print("Dataset shape:", df.shape)

print("\nMissing values per column:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
df.info()

print("\nDate range:")
print(df["Date"].min(), "to", df["Date"].max())

print("\nUnique stores:", df["Store ID"].nunique())
print("Unique products:", df["Product ID"].nunique())

print("\nSales summary:")
print(df["Units Sold"].describe())
```

```python
daily_sales = (
    df.groupby("Date", as_index=False)["Units Sold"]
      .sum()
)

fig = px.line(
    daily_sales,
    x="Date",
    y="Units Sold",
    title="Daily Total Units Sold",
    labels={
        "Date": "Date",
        "Units Sold": "Total Units Sold"
    }
)

fig.show()
```

```python
df.info()

print("\nDate range:")
print(df["Date"].min(), "to", df["Date"].max())

print("\nUnique stores:", df["Store ID"].nunique())
print("Unique products:", df["Product ID"].nunique())

print("\nSales statistics:")
print(df["Units Sold"].describe())
```

```python
category_sales = (
    df.groupby("Category", as_index=False)["Units Sold"]
      .sum()
      .sort_values("Units Sold", ascending=False)
)

category_sales
```

```python
fig = px.bar(
    category_sales,
    x="Category",
    y="Units Sold",
    title="Total Units Sold by Product Category",
    labels={"Units Sold": "Total Units Sold"},
    text="Units Sold"
)

fig.show()
```

```python
region_sales = (
    df.groupby("Region", as_index=False)["Units Sold"]
      .sum()
      .sort_values("Units Sold", ascending=False)
)

region_sales
```

```python
fig = px.bar(
    region_sales,
    x="Region",
    y="Units Sold",
    title="Total Units Sold by Region",
    text="Units Sold"
)

fig.show()
```

```python
inventory_summary = (
    df.groupby("Category", as_index=False)
      .agg(
          Average_Inventory=("Inventory Level", "mean"),
          Average_Daily_Sales=("Units Sold", "mean"),
          Total_Units_Ordered=("Units Ordered", "sum")
      )
      .round(2)
)

inventory_summary
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

actual = df["Units Sold"]
forecast = df["Demand Forecast"]

mae = mean_absolute_error(actual, forecast)
rmse = np.sqrt(mean_squared_error(actual, forecast))

print(f"MAE: {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
```

```python
# Count and inspect negative demand forecasts
negative_forecasts = df[df["Demand Forecast"] < 0]

print("Negative forecast rows:", len(negative_forecasts))
print("Percentage of dataset:", round(len(negative_forecasts) / len(df) * 100, 2), "%")

negative_forecasts[
    ["Date", "Store ID", "Product ID", "Category",
     "Units Sold", "Demand Forecast"]
].head(15)
```

```python
negative_forecasts.groupby("Category").agg(
    Negative_Forecast_Rows=("Demand Forecast", "size"),
    Average_Forecast=("Demand Forecast", "mean"),
    Minimum_Forecast=("Demand Forecast", "min")
).sort_values("Negative_Forecast_Rows", ascending=False)
```
